# Exploratory Data Analysis & Table Relationships

## Overview
To support customer service operations, the AI assistant needs to query and combine information from multiple tables:

1. **Orders & Customers (`orders` + `customers`)**:
   - The `orders` table stores purchased products and their delivery status.
   - We join `orders` with `customers` (using `customer_id`) to link each order to the respective customer's profile and contact details.

2. **Products & Promotions (`products` + `promotions`)**:
   - The `products` table is used to check product details and stock availability.
   - The `promotions` table tracks active discounts and special offers.
   - We join these two tables (using `product_id`) so the assistant can inform customers about both product availability and ongoing promotions.

## 1. Setup

```python
import pandas as pd
```

--- 
## 2. Orders & Customers

### 2.1. Inspecting the `orders` Table
Contains order records, payment methods, tracking codes, and delivery statuses.

```python
orders = pd.read_csv('../data/desafio_tecnico_ai_eng - orders.csv')
orders.head()
```

### 2.2. Inspecting the `customers` Table
Contains customer identification and contact details.

```python
customers = pd.read_csv('../data/desafio_tecnico_ai_eng - customers.csv')
customers.head()
```

### 2.3. Joining `orders` and `customers`
We perform a left join on `customer_id` to enrich each order with the customer's `name` and `email`.

```python
df_orders_with_customers = orders.merge(
    customers[["customer_id", "name", "email"]],
    on="customer_id",
    how="left"
)

df_orders_with_customers.head()
```

---
## 3. Products & Promotions

### 3.1. Inspecting the `products` Table
Contains catalog items, prices, technical specifications, and current stock quantities.

```python
products = pd.read_csv('../data/desafio_tecnico_ai_eng - products.csv')
products.head()
```

### 3.2. Inspecting the `promotions` Table
Contains discount percentages, campaign descriptions, and whether each promotion is currently active.

```python
promotions = pd.read_csv('../data/desafio_tecnico_ai_eng - promotions.csv')
promotions.head()
```

### 3.3. Joining `products` and `promotions`
We perform a left join on `product_id` to combine product stock information with any associated promotional discounts.

```python
df_products_with_promotion = products.merge(
    promotions[["product_id", "discount_percent", "description", "is_active"]],
    on="product_id",
    how="left"
)

df_products_with_promotion.head()
```
